```python
import tkinter as tk
from tkinter import messagebox
import json
import pyttsx3

# Load JSON data
def load_quiz_data(level_file):
    with open(level_file, 'r') as file:
        quiz_data = json.load(file)
    print(quiz_data)  # Confirm loading quiz data
    return quiz_data

def change_level():
    global quiz_data, current_question, score, incorrect_answers
    level = level_var.get()
    quiz_data = load_quiz_data(f'{level}_data.json')
    current_question = 0
    score = 0
    incorrect_answers = 0
    running_quiz_label.config(text=f"Currently running: {level.capitalize()} Level Quiz")
    display_question()

quiz_data = None
current_question = 0
score = 0
incorrect_answers = 0

# Initialize text-to-speech engine
engine = pyttsx3.init()

# Function to ask question and speak it
def display_question():
    global current_question
    if current_question < len(quiz_data['ques']):
        print(f"Displaying question {current_question + 1}: {quiz_data['ques'][current_question]}")  # Debugging
        question_label.config(text=quiz_data['ques'][current_question], fg="white", bg="#4CAF50", padx=20, pady=20, relief="solid", bd=3, font=("Arial", 18, "bold"))
        for i in range(4):
            options[i].config(text=quiz_data['choices'][current_question][i])
        
        # Speak the question and choices
        root.after(500, speak_question_and_choices, quiz_data['ques'][current_question], quiz_data['choices'][current_question])
    else:
        print("No more questions to display.")  # Debugging
        messagebox.showinfo("Error", "No questions available.")
        root.quit()  # Exit the app

# Function to speak the question and choices
def speak_question_and_choices(question, choices):
    engine.say(question)
    engine.say("Your choices are:")
    for i, choice in enumerate(choices):
        engine.say(f"{i + 1}. {choice}")
    engine.runAndWait()

# Function to check the answer
def check_answer():
    global current_question, score, incorrect_answers
    selected = selected_option.get()
    print(f"Selected answer: Option {selected}")
    print(f"Correct answer: {quiz_data['ans'][current_question]}")
    
    if selected == quiz_data['ans'][current_question]:
        score += 1
        print("Correct answer!")
    else:
        incorrect_answers += 1
        print("Incorrect answer.")
    
    current_question += 1
    if current_question < len(quiz_data['ques']):
        display_question()
    else:
        messagebox.showinfo("Quiz Completed", f"Your score is {score}/{len(quiz_data['ques'])}.")
        engine.say(f"Quiz completed. Your score is {score} out of {len(quiz_data['ques'])}. You got {score} correct and {incorrect_answers} incorrect.")
        engine.runAndWait()
        root.destroy()  # Properly close the application

# Create the main window
root = tk.Tk()
root.title("Voice Quiz Generator App")
root.geometry("600x500")

frame = tk.Frame(root)
frame.pack(pady=20)

level_var = tk.StringVar(value='beginner')
level_menu = tk.OptionMenu(root, level_var, "beginner", "intermediate", "advanced", command=lambda _: change_level())
level_menu.pack()

running_quiz_label = tk.Label(root, text="Currently running: Beginner Level Quiz", font=("Arial", 12), fg="blue")
running_quiz_label.pack()

question_label = tk.Label(frame, text="This is a sample question", font=("Arial", 18, "bold"), wraplength=500, fg="white", bg="#4CAF50", padx=20, pady=20, relief="solid", bd=3)
question_label.pack()

selected_option = tk.IntVar()
options = []
for i in range(4):
    radio_btn = tk.Radiobutton(frame, text=f"Option {i+1}", variable=selected_option, value=i+1, font=("Arial", 14))
    radio_btn.pack(anchor="w")
    options.append(radio_btn)

submit_btn = tk.Button(root, text="Submit", font=("Arial", 14), command=check_answer)
submit_btn.pack(pady=20)

quiz_data = load_quiz_data('beginner_data.json')
print(quiz_data)
display_question()

submit_btn.config(command=check_answer)

root.mainloop()
```
